# For Base Models

```python
# Importing the Libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier, StackingClassifier
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
from imblearn.over_sampling import RandomOverSampler, SMOTE, BorderlineSMOTE, SVMSMOTE, ADASYN
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.preprocessing import LabelEncoder
from imblearn.pipeline import Pipeline


# load the dataset
dataset = pd.read_excel('BBDM Project.xlsx')

# Prepare Features and Target
X = dataset.drop(columns='RMC', axis=1)
Y = dataset['RMC']

# Encode target variable into integer labels
label_encoder = LabelEncoder()
Y_encoded = label_encoder.fit_transform(Y)

# Split the data into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y_encoded, test_size=0.2, random_state=42)

# Data Standardization
scaler = RobustScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Define your base models
models = {
    'LR': LogisticRegression(C=10, dual=False, max_iter=100, multi_class='auto', penalty='l2', solver='sag', tol=0.01),
    'SVM': SVC(C=29.37, kernel='rbf', gamma='scale', probability=True),
    'DT': DecisionTreeClassifier(criterion='gini', max_depth=7, min_samples_leaf=9, min_samples_split=2),
    'RF': RandomForestClassifier(max_depth=10, max_features='auto', min_samples_leaf=2, min_samples_split=10, n_estimators=50),
    'KNN': KNeighborsClassifier(n_neighbors=36, weights='distance', algorithm='auto', p=1),
    'XGBoost': XGBClassifier(colsample_bytree=0.8, gamma=1, learning_rate=0.01, max_depth=7, n_estimators=50, subsample=0.8),
    'Bagging': BaggingClassifier(base_estimator=DecisionTreeClassifier(), n_estimators=100, max_samples=0.7, max_features=1.0, bootstrap=True, bootstrap_features=True)
}

# Initialize oversampling techniques
oversamplers = {
    'ROS': RandomOverSampler(random_state=42),
    'SMOTE': SMOTE(random_state=42),
    'BLSMOTE': BorderlineSMOTE(random_state=42),
    'SVMSMOTE': SVMSMOTE(random_state=42),
    'ADASYN': ADASYN(random_state=42)
}
```

```python
# Assuming the initialization of models, oversamplers, and dataset preparation steps are done before this

# Initialize a DataFrame to store results
columns = ['Oversampler', 'Model', 'Metric', 'Value']
results = pd.DataFrame(columns=columns)

# Nested loops to fill the DataFrame
for oversampler_name, oversampler in oversamplers.items():
    X_train_oversampled, Y_train_oversampled = oversampler.fit_resample(X_train_scaled, Y_train)
    for model_name, model in models.items():
        model.fit(X_train_oversampled, Y_train_oversampled)
        Y_pred = model.predict(X_test_scaled)
        
        # Metrics
        metrics = {
            'ACC': accuracy_score(Y_test, Y_pred),
            'PRE': precision_score(Y_test, Y_pred, average='weighted', zero_division=0),
            'REC': recall_score(Y_test, Y_pred, average='weighted'),
            'F1': f1_score(Y_test, Y_pred, average='weighted')
        }
        
        for metric_name, value in metrics.items():
            results = results.append({'Oversampler': oversampler_name, 'Model': model_name, 'Metric': metric_name, 'Value': value}, ignore_index=True)

```

```python
# Pivot the results to get the desired format
pivot_table = pd.pivot_table(results, values='Value', index=['Model'], columns=['Oversampler', 'Metric'], aggfunc=np.mean)

# Display the pivot table
print(pivot_table.round(2))
```
