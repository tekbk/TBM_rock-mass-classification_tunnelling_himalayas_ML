```python
# Importing the Libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import RobustScaler
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split


# load the dataset
dataset = pd.read_excel('BBDM Project.xlsx')

# Encode the target variable
encoder = LabelEncoder()
dataset['RMC'] = encoder.fit_transform(dataset['RMC'])
original_class_names = encoder.classes_  # Store original class names

# Prepare Features and Target
X = dataset.drop(columns='RMC', axis=1)
Y = dataset['RMC']

# Split into Train/Test Sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)

# Scale Features
scaler = RobustScaler()
scaler.fit(X_train)  # Fit on the training data
X_train_standardized = scaler.transform(X_train)
X_test_standardized = scaler.transform(X_test)

# Define the logistic regression model
classifier = LogisticRegression(random_state=42)

# Define the hyperparameters to tune
param_grid = {
    'penalty': ['l1', 'l2', 'elasticnet', None],
    'dual': [True, False],
    'tol': [1e-4, 1e-3, 1e-2],
    'C': [0.1, 1, 10],
    'solver': ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'],
    'max_iter': [100, 500, 1000],
    'multi_class': ['auto', 'ovr', 'multinomial']
}

# Grid Search Cross-Validation
grid_search = GridSearchCV(classifier, param_grid, cv=10, verbose=0, scoring='accuracy')

# Fit the grid search
grid_search.fit(X_train_standardized, Y_train)

# Print the best hyperparameters
print("Best Hyperparameters:", grid_search.best_params_)
```
